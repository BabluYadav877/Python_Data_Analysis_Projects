## **KPI'S**

```python
total_vehicles = len(df)
total_sales_volume = df["Sales_Volume"].sum()

total_revenue = (df["Price_USD"] * df["Sales_Volume"]).sum()

average_price = df["Price_USD"].mean()
average_mileage = df["Mileage_KM"].mean()

average_engine_size = df["Engine_Size_L"].mean()

best_selling_model = (
    df.groupby("Model")["Sales_Volume"]
    .sum()
    .idxmax()
)

region_revenue = (
    df.groupby("Region")
    .apply(lambda x: (x["Price_USD"] * x["Sales_Volume"]).sum())
)

top_revenue_region = region_revenue.idxmax()

print("========== BMW BUSINESS KPIs ==========")
print(f"1.Total Vehicles: {total_vehicles:,}")
print(f"2.Total Sales Volume: {total_sales_volume:,}")
print(f"3.Total Revenue: ${total_revenue:,.2f}")
print(f"4.Average Price: ${average_price:,.2f}")
print(f"5.Average Mileage: {average_mileage:,.2f} KM")
print(f"6.Average Engine Size: {average_engine_size:.2f} L")
print(f"7.Best Selling Model: {best_selling_model}")
print(f"8.Top Revenue Region: {top_revenue_region}")

```

 **Create Revenue Column**


```python
df["Revenue"] = df["Price_USD"] * df["Sales_Volume"]
```

## **Visuals Chart**

**1️⃣ Correlation Heatmap**

```python
plt.figure()
numeric_cols = [
    "Price_USD", "Mileage_KM", "Engine_Size_L",
    "Sales_Volume", "Composite_Health_Score",
    "Market_Saturation_Ratio", "Revenue"
]

corr = df[numeric_cols].corr()

plt.imshow(corr)
plt.title("Correlation Heatmap")
plt.xticks(range(len(numeric_cols)),numeric_cols,rotation=45)
plt.yticks(range(len(numeric_cols)),numeric_cols)
plt.colorbar()
plt.show()

```

**2️⃣ Sales Trend by Year (Line Chart)**

```python
year_sales = df.groupby("Year")["Sales_Volume"].sum()

plt.figure()
plt.plot(year_sales.index, year_sales.values)
plt.title("Total Sales volume By Year")
plt.xlabel("Year")
plt.ylabel("Sales Volume")
plt.xticks(rotation=45)
plt.show()
```

**3️⃣ Revenue by Region (Bar Chart)**

```python
region_revenue = df.groupby("Region")["Revenue"].sum()

plt.figure()
plt.bar(region_revenue.index, region_revenue.values)
plt.title("Total Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

**4️⃣ Price Distribution (Histogram)**

```python
plt.figure()
plt.hist(df["Price_USD"], bins=30)
plt.title("Price Distribution")
plt.xlabel("Price (USD)")
plt.ylabel("Frequency")
plt.show()
```

**5️⃣ Mileage vs Price (Scatter + Regression Line)**

```python
df["Price_Category"] = pd.cut(
    df["Price_USD"],
    bins=[0, 50000, 80000, 120000],
    labels=["Budget", "Mid-Range", "Premium"]
)

price_counts = df["Price_Category"].value_counts()

plt.figure()
plt.pie(price_counts, labels=price_counts.index, autopct='%1.1f%%')
plt.title("Price Distribution by Category")
plt.show()
```

**Pie Chart of Revenue by Mileage Category**

```python
df["Mileage_Category"] = pd.cut(
    df["Mileage_KM"],
    bins=[0, 50000, 100000, 150000, 200000],
    labels=["Low (<50K)", "Medium (50K-100K)", "High (100K-150K)", "Very High (>150K)"]
)

# Revenue by Mileage Category
revenue_by_mileage = df.groupby("Mileage_Category")["Revenue"].sum()

# Pie Chart
plt.figure()
plt.pie(revenue_by_mileage, labels=revenue_by_mileage.index, autopct='%1.1f%%')
plt.title("Revenue Share by Mileage Category")
plt.show()
```

**6️⃣ Engine Size vs Price (Scatter + Regression)**

```python
plt.figure()
plt.scatter(df["Engine_Size_L"], df["Price_USD"])

z = np.polyfit(df["Engine_Size_L"], df["Price_USD"], 1)
p = np.poly1d(z)
plt.plot(df["Engine_Size_L"], p(df["Engine_Size_L"]))

plt.title("Engine Size vs Price (with Regression)")
plt.xlabel("Engine Size (L)")
plt.ylabel("Price (USD)")
plt.show()
```

**7️⃣ Fuel Type Distribution**

```python
fuel_counts = df["Fuel_Type"].value_counts()

plt.figure()
plt.bar(fuel_counts.index, fuel_counts.values)
plt.title("Fuel Type Distribution")
plt.xlabel("Fuel Type")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

**8️⃣ Cluster Distribution (KMeans)**

```python
cluster_counts = df["Cluster_KMeans"].value_counts()

plt.figure()
plt.bar(cluster_counts.index.astype(str), cluster_counts.values)
plt.title("Customer Cluster Distribution")
plt.xlabel("Cluster")
plt.ylabel("Count")
plt.show()
```

## **VISUALIZATIONS**

```python
gender_count = df['gender'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(gender_count,labels=gender_count.index,autopct='%1.1f%%')
plt.title("Gender Distribution")
plt.show()
```

```python
job = df['part_time_job'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(job,labels=job.index,autopct='%1.1f%%')
circle = plt.Circle((0,0),0.70,fc='white')
plt.gca().add_artist(circle)
plt.title("Students Having Part-Time Job")
plt.show()
```

```python
level_exam = df.groupby('academic_level')['exam_score'].mean()

level_exam.plot(kind='bar')
plt.title("Academic Level vs Exam Score")
plt.ylabel("Exam Score")
plt.show()
```

```python
internet_prod = df.groupby('internet_quality')['productivity_score'].mean()

internet_prod.plot(kind='barh')
plt.title("Internet Quality vs Productivity")
plt.show()
```

```python
stack_bar = df.groupby('academic_level')[['study_hours','sleep_hours']].mean()

stack_bar.plot(kind='barh', stacked=True)

plt.title("Academic Level vs Study Hours & Sleep Hours")
plt.xlabel("Average Hours")
plt.ylabel("Academic Level")

plt.show()
```

```python
stack_col = df.groupby('internet_quality')[['productivity_score','mental_health_score']].mean()

stack_col.plot(kind='bar', stacked=True)

plt.title("Internet Quality vs Productivity & Mental Health")
plt.xlabel("Internet Quality")
plt.ylabel("Score")

plt.show()
```

```python
plt.scatter(df['burnout_level'],df['exam_score'])
plt.title("Burnout vs Exam Score")
plt.xlabel("Burnout Level")
plt.ylabel("Exam Score")
plt.show()
```
